## 3. Strings

Strings are sequences of characters. Python uses zero-based indexing.

````python
word = "Python"

print(word[0])
print(word[1])
print(word[-1])
print(len(word))
print(word.upper())
print(word.lower())
````

## 4. Numbers and Operators

Common operators: `+`, `-`, `*`, `/`, `//`, `%`, `**`.

````python
a = 10
b = 3

print(a + b)
print(a - b)
print(a * b)
print(a / b)
print(a // b)
print(a % b)
print(a ** 2)
````

## 5. Comparison Operators

Comparisons produce `True` or `False`.

`==` equal, `!=` not equal, `>`, `<`, `>=`, `<=`

````python
age = 19

print(age >= 18)
print(age == 20)
print(age != 15)
````

## 6. Lists

A list stores multiple values in an ordered collection. Lists are **mutable**, so their contents can be changed.

````python
scores = [75, 82, 91, 68, 88]

print(scores)
print(scores[0])
print(scores[-1])
print(len(scores))
````

### Changing and adding list elements

````python
scores = [75, 82, 91, 68, 88]

scores[0] = 80
scores.append(95)

print(scores)
````

### Adding and removing elements

Useful methods include `append()`, `insert()`, `remove()` and `pop()`.

````python
numbers = [10, 20, 30]

numbers.append(40)
numbers.insert(1, 15)
print(numbers)

numbers.remove(20)
print(numbers)

removed = numbers.pop()
print("Removed:", removed)
print(numbers)
````

## 7. Searching in a List

Use `in` to check whether a value exists.

````python
students = ["Ali", "Sara", "Mehmet", "Ayse"]

print("Sara" in students)
print("John" in students)
````

Use `index()` to find the position of an existing value.

````python
students = ["Ali", "Sara", "Mehmet", "Ayse"]

print(students.index("Mehmet"))
````

````python
a = 40
b = 60
if (a > b):
   print("a is greater")
else:
   print("b is greater")
````

### Searching using a condition

Loops are useful when you want to inspect every element.

````python
scores = [45, 78, 91, 52, 88]

for score in scores:
    if score >= 80:
        print("High score:", score)
````

````python
a = int(input("Enter an integer"))
flag = 1
for i in range(2, a):
  if(a % i == 0):
    print("Not prime")
    flag = 0
    break

if(flag == 1):
  print("Prime")
````

## 8. Sorting a List

- `sort()` changes the original list.
- `sorted()` creates a new sorted list.

````python
numbers = [40, 10, 80, 30, 60]

numbers.sort()
print(numbers)
````

````python
numbers = [40, 10, 80, 30, 60]

sorted_numbers = sorted(numbers)

print("Original:", numbers)
print("Sorted:", sorted_numbers)
````

### Descending order

````python
numbers = [40, 10, 80, 30, 60]

numbers.sort(reverse=True)
print(numbers)
````

## 9. Indexing and Slicing

The general form is `sequence[start:stop]`. The `stop` position is not included.

````python
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[1:4])
print(numbers[:3])
print(numbers[3:])
print(numbers[::2])
print(numbers[::-2])
````

## 11. Sets

A set stores **unique values**. Sets are useful for removing duplicates and for set operations.

````python
numbers = [10, 20, 10, 30, 20, 40]

unique_numbers = set(numbers)
print(unique_numbers)
````

## 15. Python List vs. NumPy Array

A Python list repeats when multiplied by an integer:

````python
numbers = [10, 20, 30]
print(numbers * 2)
````

A NumPy array performs numerical multiplication:

````python
import numpy as np

numbers = np.array([10, 20, 30])
print(numbers * 2)
````

### Key idea

Python list:

`[10, 20, 30] * 2` → `[10, 20, 30, 10, 20, 30]`

NumPy array:

`np.array([10, 20, 30]) * 2` → `[20, 40, 60]`

This difference is extremely important in data science.

## 16. NumPy Indexing, Slicing and Searching

### Sorting a NumPy array

````python
import numpy as np

scores = np.array([75, 82, 91, 68, 88])

print(np.sort(scores))

scores.sort()
print(scores)
````

## 17. Multidimensional NumPy Arrays

A 2D array can represent a table with rows and columns.

````python
import numpy as np

data = np.array([
    [80, 75, 90],
    [65, 88, 72],
    [92, 95, 89]
])

print(data)
print("Shape:", data.shape)
print("Rows:", data.shape[0])
print("Columns:", data.shape[1])
````

````python
print("First row:", data[0])
print("First column:", data[:, 0])
print("First two rows:", data[:2])
````

## 21. Functions

Functions organize reusable code. They can receive inputs and return a result.

````python
def calculate_average(a, b, c):
    return (a + b + c) / 3

average = calculate_average(80, 90, 70)
print(average)
````

## 22. Useful Built-in Functions

Common functions include `len()`, `min()`, `max()`, `sum()`, and `sorted()`.

````python
numbers = [10, 20, 30, 40, 50]

print("Length:", len(numbers))
print("Minimum:", min(numbers))
print("Maximum:", max(numbers))
print("Sum:", sum(numbers))
print("Sorted:", sorted(numbers))
````

## 23. List Comprehension

A list comprehension is a concise way to create a list.

````python
squares = []

for x in range(1, 6):
    squares.append(x ** 2)

print(squares)
````

````python
squares = [x ** 2 for x in range(1, 6)]
print(squares)
````

## 25. A Small Data Science Example

Suppose these are exam scores. Use NumPy to calculate simple statistics.

````python
import numpy as np

scores = np.array([65, 72, 88, 91, 55, 78, 84, 93, 69, 87])

print("Average:", np.mean(scores))
print("Highest:", np.max(scores))
print("Lowest:", np.min(scores))
print("Median:", np.median(scores))
````

````python
high_scores = scores[scores > 80]

print("Scores above 80:", high_scores)
print("Number of high scores:", len(high_scores))
````

## 26. A First Look at Pandas

Pandas provides convenient structures for working with tabular data.

A `DataFrame` is similar to a spreadsheet or database table.

````python
import pandas as pd

data = {
    "Name": ["Ali", "Sara", "Mehmet", "Ayse"],
    "Age": [19, 20, 19, 21],
    "Score": [85, 92, 76, 88]
}

df = pd.DataFrame(data)
print(df)
````

````python
print(df["Score"])
print("Average:", df["Score"].mean())
print(df[df["Score"] >= 80])
````
