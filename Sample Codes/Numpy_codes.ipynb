# Define methods and sort arrays

```python
import numpy as np
def selection_sort(x):
   for i in range(len(x)):
      swap = i + np.argmin(x[i:])
      (x[i], x[swap]) = (x[swap], x[i])
   return x
```

```python
x = np.array([2, 1, 4, 3, 5])
selection_sort(x)
```

```python
x = np.array([2, 1, 4, 3, 5])
i = np.argsort(x)
print(i)
```

**Cosine Similarity**

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity

# Sample customer reviews
review1 = "I love this phone. The camera is amazing and battery lasts long."
review2 = "This phone has a great camera and long battery life. I really like it."

# Convert texts to word count vectors
vectorizer = CountVectorizer().fit([review1, review2])
vectors = vectorizer.transform([review1, review2])

# Calculate cosine similarity between the two vectors
cos_sim = cosine_similarity(vectors[0], vectors[1])

print(f"Cosine Similarity between reviews: {cos_sim[0][0]:.3f}")
```

```python
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np
import pandas as pd

# Sample documents
documents = [
    "Machine learning is fascinating and powerful.",
    "Deep learning is a subset of machine learning.",
    "Artificial intelligence and machine learning are closely related.",
    "I love watching movies on weekends.",
    "The movie was thrilling and exciting."
]

# Step 1: Convert documents to a document-term matrix (word-document coincidence matrix)
vectorizer = CountVectorizer()
doc_term_matrix = vectorizer.fit_transform(documents)

# Get feature (word) names
words = vectorizer.get_feature_names_out()

# Convert sparse matrix to dense and create a DataFrame for easy viewing
df_doc_term = pd.DataFrame(doc_term_matrix.toarray(), columns=words)
df_doc_term.index = [f'Doc {i+1}' for i in range(len(documents))]

print("Word-Document Coincidence Matrix (Document-Term Matrix):")
print(df_doc_term)

# Step 2: Calculate cosine similarity matrix between all documents
similarity_matrix = cosine_similarity(doc_term_matrix)

# Convert similarity matrix to DataFrame for better display
df_similarity = pd.DataFrame(similarity_matrix, columns=df_doc_term.index, index=df_doc_term.index)

print("\nCosine Similarity Matrix:")
print(df_similarity.round(3))

# Step 3: Find the pair of documents with the highest similarity (excluding self-similarity)
np.fill_diagonal(similarity_matrix, 0)  # Ignore similarity of document with itself
most_similar_pair = np.unravel_index(similarity_matrix.argmax(), similarity_matrix.shape)

doc1_index, doc2_index = most_similar_pair
print(f"\nMost similar documents are Document {doc1_index + 1} and Document {doc2_index + 1}")
print(f"Document {doc1_index + 1}: {documents[doc1_index]}")
print(f"Document {doc2_index + 1}: {documents[doc2_index]}")
print(f"Cosine Similarity: {similarity_matrix[doc1_index, doc2_index]:.3f}")
```

**Object Similarity example**

```python
def numeric_similarity(x, y, min_val, max_val):
    return 1 - abs(x - y) / (max_val - min_val)

def binary_similarity(x, y):
    return 1.0 if x == y else 0.0

def nominal_similarity(x, y):
    return 1.0 if x == y else 0.0

def ordinal_similarity(x, y, min_rank, max_rank):
    return 1 - abs(x - y) / (max_rank - min_rank)

# -----------------------------
# Example data
ideal = {
    "rent": 1000,          # numeric
    "balcony": 1,          # binary
    "neighborhood": "Central",  # nominal
    "furnishing": 3        # ordinal rank (1=unfurnished, 2=partial, 3=full)
}

apartment_A = {
    "rent": 900,
    "balcony": 1,
    "neighborhood": "Central",
    "furnishing": 3
}

apartment_B = {
    "rent": 1100,
    "balcony": 0,
    "neighborhood": "Suburban",
    "furnishing": 2
}

# -----------------------------
# Parameters for normalization
rent_min, rent_max = 800, 1200
furnish_min, furnish_max = 1, 3

def overall_similarity(obj, ideal):
    sims = []
    sims.append(numeric_similarity(obj["rent"], ideal["rent"], rent_min, rent_max))
    sims.append(binary_similarity(obj["balcony"], ideal["balcony"]))
    sims.append(nominal_similarity(obj["neighborhood"], ideal["neighborhood"]))
    sims.append(ordinal_similarity(obj["furnishing"], ideal["furnishing"], furnish_min, furnish_max))
    return sum(sims) / len(sims)

# -----------------------------
# Compute similarities
sim_A = overall_similarity(apartment_A, ideal)
sim_B = overall_similarity(apartment_B, ideal)

print(f"Similarity of Apartment A to Ideal: {sim_A:.4f} ({sim_A*100:.2f}%)")
print(f"Similarity of Apartment B to Ideal: {sim_B:.4f} ({sim_B*100:.2f}%)")
```
